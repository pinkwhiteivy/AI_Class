```python
import matplotlib.pyplot as plt
from sklearn import datasets, metrics
from sklearn.model_selection import train_test_split

digits  = datasets.load_digits()
plt.imshow(digits.images[0], cmap=plt.cm.gray_r, interpolation='nearest')

n_samples = len(digits.images)
data = digits.images.reshape((n_samples, -1))

from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=6)

X_train, X_test, y_train, y_test = train_test_split(
    data, digits.target, test_size = 0.2)

knn.fit(X_train, y_train)

y_pred = knn.predict(X_test)

scores = metrics.accuracy_score(y_test, y_pred)
print(scores)

plt.imshow(X_test[10].reshape(8,8), cmap=plt.cm.gray_r, interpolation='nearest')
y_pred = knn.predict([X_test[10]])
print(y_pred)
```

```python
from sklearn.tree import DecisionTreeClassifier

dt = DecisionTreeClassifier(random_state=42)
dt.fit(X_train, y_train)

y_pred_dt = dt.predict(X_test)
scores_dt = metrics.accuracy_score(y_test, y_pred_dt)
print(f"Decision Tree 정확도: {scores_dt}")

single_pred_dt = dt.predict([X_test[10]])
print(f"예측값: {single_pred_dt}")
```

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression(max_iter=5000, random_state=42)
lr.fit(X_train, y_train)

y_pred_lr = lr.predict(X_test)
scores_lr = metrics.accuracy_score(y_test, y_pred_lr)
print(f"Logistic Regression 정확도: {scores_lr}")

single_pred_lr = lr.predict([X_test[10]])
print(f"예측값: {single_pred_lr}")
```

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(random_state=42)
rf.fit(X_train, y_train)

y_pred_rf = rf.predict(X_test)
scores_rf = metrics.accuracy_score(y_test, y_pred_rf)
print(f"Random Forest 정확도: {scores_rf}")

single_pred_rf = rf.predict([X_test[10]])
print(f"예측값: {single_pred_rf}")
```
